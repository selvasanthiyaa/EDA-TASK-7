```python
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for col in categorical_cols:
  df[col] = (
      df[col]
      .astype('string')
      .str.strip()
      .str.lower()
  )

  for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
score_cols = [
    'math score',
    'reading score',
    'writing score',
]

stats = df[score_cols].agg(
    ['mean', 'median', 'std']
).T

stats['Q1'] = df[score_cols].quantile(0.25)
stats['Q2'] = df[score_cols].quantile(0.50)
stats['Q3'] = df[score_cols].quantile(0.75)

print(stats.round(2))
```

```python
df['total mark'] = (
    df['math score']
    + df['reading score']
    + df['writing score']
)

df['percentage'] = (
    df['total mark'] / 300
) * 100

print(df[[
    'math score', 'reading score', 'writing score',
    'total mark', 'percentage'
]].head())

print(df[['total mark', 'percentage']].describe())
```

```python
for col in score_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, "Outliers:", len(outliers))

# Save the cleaned dataset
df.to_csv("Student_Performance_Cleaned.csv", index=False)

print("Task completed!")
```
